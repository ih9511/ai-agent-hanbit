```python
from langchain_tavily import TavilySearch


tool = TavilySearch(max_results=2)
tool.invoke({"query": "랭그래프가 무엇인가요?"})
```

```python
# 랭체인 기반 도구 만들기
# 직접 덧셈과 곱셈 도구 만들기
from langchain.tools import tool

@tool
def add(a: int, b: int) -> int:
    """Adds a and b.
    
    Args:
        a: first int
        b: second int    
    """
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiplies a and b.
    
    Args:
        a: first int
        b: second int
    """
    return a * b

tools = [add, multiply]
    
```

```python
# 도구 호출을 위한 bind_tools 사용하기
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o")
llm_with_tools = llm.bind_tools(tools)
```

```python
# 도구의 사용이 필요한 질문 입력하기
query = "3 곱하기 5는 뭔가요? 그리고 2 더하기 4는 뭔가요?"

response = llm_with_tools.invoke(query)
response
```

```python
# tool_calls로 도구 호출 결과 확인하기
response.tool_calls
```

```python
# 도구의 사용이 필요 없는 질문 입력하기
query = "안녕하세요"

response = llm_with_tools.invoke(query)
response
```

```python
response.tool_calls
```

```python
# 타빌리 서치 도구를 바인딩한 모델 사용하기
# 웹 검색 도구를 사용한 LLM 만들기
from langchain_openai import ChatOpenAI

tool = TavilySearch(max_results=2)
tools = [tool]

llm = ChatOpenAI(model="gpt-4o")
llm_with_tools = llm.bind_tools(tools)
```

```python
llm_with_tools.invoke("안녕")
```

```python
response = llm_with_tools.invoke("2026년 AI 트렌드는?")
response
```
