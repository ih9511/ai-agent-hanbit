## 6.3.1 사용자 도구 정의하기

- 도구를 정의하는 가장 간단한 방법은 `@tool` 데코레이터를 사용하는 것.
  
1. 도구의 이름: 도구의 이름은 기본적으로 함수명을 사용함. 만약 함수명이 아닌 새로운 이름을 정의하고 싶다면, 데코레이터의 첫 번째 인자로 문자열을 입력할 수 있음.
```python
@tool
def my_tool():
    ...
```
```python
@tool("도구의 이름")
```
2. 도구의 기능: 함수 내 docstring을 통해 도구의 기능에 대한 설명을 작성할 수 있음. **에이전트가 작업에 필요한 도구를 호출할 때 이 docstring 설명을 참고**하므로 명확하게 작성할 필요가 있음.

````python
from langchain.tools import tool


@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b

print(multiply.name)
print(multiply.description)
print(multiply.args)
````

3. Argument 스키마: 도구의 입력으로 사용될 argument의 경우 데코레이터의 args_schema 속성을 사용하여 정의할 수 있음.

````python
from pydantic import BaseModel, Field
from langchain.tools import tool


class CalculatorInput(BaseModel):
    a: int = Field(description="first number")
    b: int = Field(description="second number")
    
@tool(args_schema=CalculatorInput)
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b

print(multiply.name)
print(multiply.description)
print(multiply.args)
````

cf.) 구글 스타일 docstring
- 한 줄 요약 (Summary line): 함수/메서드가 어떤 역할을 하는지를 간결하게 설명한 문장으로 시작.
- 상세 설명 (Optional Detailed description): 필요에 따라 함수의 동작이나 맥락, 주의 사항 등을 추가로 설명.
- Args 섹션: 파라미터 각각에 대해 이름과 간단한 설명을 제공함. 각 줄은 두 칸 혹은 네 칸 들여쓰기 후 "파라미터명: 설명" 형식으로 작성함.
- Returns 섹션: 반환값이 무엇인지 명시.

````python
# 구글 스타일 docstring
@tool(parse_docstring=True)
def multiply(a: int, b: int) -> int:
    """Multiply two numbers.
    
    Args:
        a: The first number.
        b: The second number.
    """
    return a * b

print(multiply.name)
print(multiply.description)
print(multiply.args)
````
